```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, to_date, lpad, collect_list, size

# ====================================================
# 1. KHỞI TẠO SPARK KẾT NỐI VỚI CLUSTER
# ====================================================
print("🚀 Đang khởi động Spark kết nối Cluster (Job Nặng)...")
spark = SparkSession.builder \
    .master("spark://spark-master:7077") \
    .appName("Cluster_Process_Full_Transactions") \
    .config("spark.hadoop.fs.defaultFS", "hdfs://namenode:9000") \
    .config("spark.driver.memory", "3g") \
    .config("spark.executor.memory", "3g") \
    .config("spark.hadoop.dfs.client.use.datanode.hostname", "true") \
    .config("spark.sql.shuffle.partitions", "200") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")
print("✅ Đã kết nối Spark thành công với 8GB RAM được cấp!")
```

```python
# ====================================================
# 2. KHAI BÁO ĐƯỜNG DẪN HDFS
# ====================================================
INPUT_PATH = "hdfs://namenode:9000/data/raw/"
OUTPUT_PATH = "hdfs://namenode:9000/data/processed/"

print("⏳ Đang đọc toàn bộ file transactions_train.csv từ HDFS...")
transactions = spark.read.csv(INPUT_PATH + "transactions_train.csv", header=True)
print("✅ Đọc file thành công!")
```

```python
# ====================================================
# 3. CHUẨN HÓA & CACHE DỮ LIỆU
# ====================================================
print("⚙️ Đang chuẩn hóa định dạng ngày tháng và ID sản phẩm...")

# Ép kiểu ngày tháng
transactions = transactions.withColumn("t_dat_date", to_date(col("t_dat"), "yyyy-MM-dd"))

# KỸ THUẬT QUAN TRỌNG: Thêm số 0 vào đầu article_id để khớp file ảnh
transactions = transactions.withColumn("article_id", lpad(col("article_id").cast("string"), 10, "0"))

print("⏳ Đang phân bổ tải (Repartition) và đẩy vào RAM (Cache)... (Sẽ mất 1-2 phút)")
# Repartition để chia đều tải ra các Worker
trans_recent = transactions.repartition(200)
trans_recent.cache()

# Gọi lệnh action (count) để ép Spark thực thi ngay quá trình Cache
total_rows = trans_recent.count()
print(f"✅ Hoàn tất! Tổng số giao dịch đang nằm trong RAM: {total_rows:,} dòng")
```

```python
# ====================================================
# 4. TẠO GIỎ HÀNG (MARKET BASKETS)
# ====================================================
print("🛒 Đang gom nhóm tạo Giỏ hàng (Market Baskets) cho Thuật toán FP-Growth...")

# Gom những món đồ được mua bởi CÙNG 1 người trong CÙNG 1 ngày thành 1 giỏ
baskets_df = trans_recent.groupBy("customer_id", "t_dat_date") \
    .agg(collect_list("article_id").alias("items")) \
    .filter(size(col("items")) > 1) # Chỉ lấy giỏ hàng có từ 2 món trở lên

baskets_path = OUTPUT_PATH + "market_baskets.parquet"
print(f"⏳ Đang ghi ra HDFS tại: {baskets_path}")

baskets_df.write.mode("overwrite").parquet(baskets_path)
print("✅ Đã lưu thành công Market Baskets!")
```

```python
# ====================================================
# 5. LƯU BẢN CLEAN & DANH SÁCH ITEM HỢP LỆ
# ====================================================
print("💾 Đang lưu bản Clean của Transactions & Valid Articles...")

# 1. Lưu bản Clean của toàn bộ Transactions (Dùng cho Bắt Trend)
clean_path = OUTPUT_PATH + "cleaned_transactions.parquet"
trans_recent.select("t_dat_date", "customer_id", "article_id", "price") \
    .write.mode("overwrite").parquet(clean_path)
print(f"✅ Đã lưu Cleaned Transactions tại: {clean_path}")

# 2. Xuất danh sách Item hợp lệ (Đẻ ra cái file valid_articles.csv mà bảng Articles đang cần chờ)
valid_path = OUTPUT_PATH + "valid_articles.csv"
trans_recent.select("article_id").distinct() \
    .write.mode("overwrite").option("header", "true").csv(valid_path)
print(f"✅ Đã lưu Valid Articles tại: {valid_path}")

print("\n🎉 XONG! TOÀN BỘ 3 KHỐI DỮ LIỆU ĐẦU RA ĐÃ NẰM AN TOÀN TRÊN HDFS.")
```

```python
# ====================================================
# 6. ĐÓNG KẾT NỐI VÀ XẢ RAM
# ====================================================
trans_recent.unpersist() # Thả bộ nhớ Cache 31.7 triệu dòng
spark.stop()             # Tắt Spark
print("🛑 Đã xả RAM và ngắt kết nối Spark thành công.")
```
